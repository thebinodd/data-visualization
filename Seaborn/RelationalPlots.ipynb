```python
import seaborn as sns
import plotly.express as px
```

```python
#Relational Plots
# used to see statistical relation between 2 or more variables
# bivariable analysis
```

```python
#Axis level Graphs -> control over graph only -> handle only one graph
#Figure Level Graphs -> control over whole figure -> handle multiple graphs
```

```python
tips = sns.load_dataset("tips") #inbuilt dataset in seaborn
tips.sample(5)
```

```python
# 1. Scatter Plots 
```

```python
#    i> Axis level Plots
sns.scatterplot(data=tips , x="total_bill" , y="tip")
```

```python
# ii> Figure level Graphs

sns.relplot(data=tips,
            x="total_bill",
            y="tip",
            kind="scatter", 
            hue="sex", #Different colors for different sex
            style="time", # Different styles for different times
            size="size" #Bigger size for bigger team
            
           )
```

```python
#2. Line Plots

gap = px.data.gapminder() #Dataframe of lifeExp , population , perCapita etc
gap.sample(5)

# Axis Level Line plots
sns.lineplot(data = gap[gap["country"] == "Nepal"],
            x="year",
             y="lifeExp"
            
            )

```

```python
#Figure Level Line plots

sns.relplot(data = gap[gap["country"] == "Nepal"] ,
           kind="line",
            x="year",
            y="lifeExp"
           )
```

```python
#Multiple Line Plots in single Figure

new_df = gap[gap["country"].isin(["Nepal" , "Germany" , "Nigeria" , "Brazil" , "India"])]
sns.relplot(
    data = new_df,
    kind="line",
    x="year",
    y="lifeExp",
    hue="country", #Separete lines for different countries
    style="continent" #Separate Styled lines for different continents
)
```

```python
# 3. Facet Plots
# -> Works only in figure level functions
# -> Don't work with scatterplot or lineplot
# -> create multiple graphs in same figure 
sns.relplot(
    data = tips,
    x="total_bill",
    y="tip",
    kind="scatter",
    col="sex", #Made side by side plot for male tips and female tips
    row = "smoker" # Make 4 graphs for male smoker , male non smoker , fem smoker and fem non smoker
    
)
```

```python
sns.relplot(
    data = gap,
    x="lifeExp",
    y="gdpPercap",
    kind="scatter",
    col = "year"
)
#The graphs are not readable since all are in same rows
```

```python
sns.relplot(
    data = gap,
    x="lifeExp",
    y="gdpPercap",
    kind="scatter",
    col = "year",
    col_wrap=3
)
#The graphs are  readable one row contains only 3 graphs
```

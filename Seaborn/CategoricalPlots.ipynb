```python
import seaborn as sns
```

```python
iris = sns.load_dataset("iris")
iris.sample(3)
```

```python
tips = sns.load_dataset("tips")
tips.sample(3)
```

```python
#------Strip Plots------

# Strip Plot shows the scatter plot in verticle line according to category


# Axis Level Plot

# sns.stripplot(data=tips ,
#               x="day" ,
#               y="total_bill",
#               jitter=False #Makes the gap between dots zero
#              )

# Figure Level Function

sns.catplot(
    kind="strip",
    data = tips,
    x="day" ,
    y="total_bill",
    jitter=False #Makes the gap between dots zero
    
)
```

```python
#------Swarm Plot------

# Same as strip plot but has better representation

sns.catplot(
    data=tips,
    kind="swarm", #Runs the internal algorithm that shows the distribution of data too
    x="day",
    y="total_bill",
    hue="sex",
)
```

```python
#-------Box Plots------

# display the distribution of data using five number summary -> lower quartile , median , higher quartile , lowest value and highest value


#Axis Level Function
sns.boxplot(
    data = tips,
    x="sex", #make separate plots for male and female
    y="total_bill" #Box Plot for numeric column
)
```

```python
# Figure Level Function

sns.catplot(
    kind="box",
    data=tips,
    y="total_bill", #Single Box Plot
)
```

```python
#------Bar Plots ------

sns.barplot(
    data = tips,
    x="sex",
    y="total_bill"
    
)

#Default it shows the mean of numeric columns

#The bar at the top is called error bar which is calculated using bootstrapping and shows the confidence interval
```

```python
sns.catplot(
    kind="bar",
    data = tips,
    x="sex",
    y="total_bill",
    hue="smoker",
    errorbar=None, #Remove error bar
    estimator="std" #Calculate standard deviation
)
```

```python
#----- Violin Plots ------
#Violin plot is the combination of KDE plot and the  box plot

sns.violinplot(
    data = tips,
    x="day",
    y="total_bill"
)

#The KDE plot is symmetrical on both side and inside this there is box plot
```

```python
sns.catplot(
    kind="violin",
    data = tips,
    y="total_bill",
    hue="sex",
    split=True
)

#Now the graph is assymmertrical which show male on one side and female in another side
```

```python
#------Point Plots-----
# Point plot is the bar plot where there is no bar instead the error bars are connected using lines
#it is majorly used to see the difference between two category
sns.pointplot(
    data = tips,
    x="sex",
    y="total_bill"
)
```

```python
#Count plots
#Used to count the number of repeatation of each category

sns.countplot(
    data = tips,
    x="sex", 
)
```

```python
#Faceting 
sns.catplot(
    data = tips,
    x="sex",
    kind="box",
    y="total_bill",
    col="day",
    col_wrap=2
)
```

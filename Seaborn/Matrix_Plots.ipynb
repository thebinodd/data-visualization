```python
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
gap = px.data.gapminder()
gap.sample(5)
```

```python
# -----Heat Maps-----
new_df = gap.pivot(index = "country" , columns = "year" , values = "lifeExp").head(30)
sns.heatmap(data = new_df)
```

```python
asia = gap[gap["continent"] == "Asia"].pivot(index = "country" , columns = "year" , values="lifeExp")
plt.figure(figsize=(15,15))
sns.heatmap(asia ,
            annot=True, #Give the numbers over the colors too
            cmap="viridis" #Use the viridis color theme
           ) 
```

```python
#----- Cluster Plots-----

#Cluster maps is similar to heatmap but cluster map shows the similarities between different columns
df = px.data.iris()
sns.clustermap(df.iloc[:,[0,1,2,3]])
```

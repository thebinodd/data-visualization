```python
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
tips.sample(3)
```

```python
#-----Regression Plots------

#Regplot -> axis level
sns.regplot(
    data = tips,
    x="total_bill",
    y="tip"
)

# Here the line is made in such way that all points are in the minimum distance from line
# The equation of the line is tip = slope * total_bill + y-intercept
# With this equation we can predict the tip if total_bill is given
# Because all the points are not touching the line , they have some error 
```

```python
# Lmplot -> Figure level
#Same as Regplot

sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

```python
# ResidPlot
# while the regression plot gives some error because not all points are touching line resid plot gives the exact distance of those points to line

sns.residplot(
    data = tips,
    x="total_bill",
    y="tip"
)

#The points which lie exactly in the line gives the correct prediction
```

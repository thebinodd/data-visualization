```python
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
tips.sample(3)
```

```python
# FacedGrid are used to plot the different graphs in one figure

#Method 1

sns.relplot(

    data = tips,
    x="total_bill",
    y="tip",
    kind="scatter",
    col="day", #make separate plots for 4 days
    row="time" #make two plot for each days
    
)
```

```python
 # Method 2

g = sns.FacetGrid(
    data = tips,
    col = "day", # makes the empty grid of 4 * 2 and store in g
    row="time",
    hue = "sex"
)

g.map(
    sns.scatterplot , "total_bill" , "tip"
)
g.add_legend()
```

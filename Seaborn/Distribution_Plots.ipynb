```python
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
# 1. Histograms
# Histograms gives univariate analysis with multiple numerical data in bins of certain widths


#------Axes level Histogram-----
sns.histplot(
    data = tips,
    x="tip"
)

```

```python
#-----Figure Level Histograms-----

sns.displot(
    data = tips,
    x="tip",
    kind="hist",
    hue="sex", #Show the different color for different sex
    element="step", #better styles for hue
    bins = 20, #Bins gap/width
)
```

```python
#We can plot categorical histogram but techincally this cannot be considered as a histogram
#This is count more than hist
sns.displot(
    data = tips,
    x="day",
    kind="hist"
)
```

```python
#Bivariate Histograms

sns.histplot(
    data = tips,
    x="total_bill",
    y="tip"
)

#Darker color = more value of tips and total_bill both
```

```python
# 2. KDE Plots
# KDE (Kernel Density Estimation) Plot show smooth line rather than bins which shows the probability/density of data
# We can find the exact value of 6,7,8.. in KDE plot since this is discrete and continuous 
# We cannot fing the exact value of 6,7,8... in Histogram since they fall in common bins

sns.kdeplot(
    data = tips,
    x="total_bill"
)
```

```python
#Figure level
sns.displot(
    data = tips,
    x="total_bill",
    kind="kde",
    hue = "sex",
    fill=True
) 
```

```python
#Bivariate KDE Plot
sns.displot(
    data = tips,
    x="tip",
    y="total_bill",
    kind="kde",

)

#Closer the circle larger the value of tips and total_bill
```

```python
# 3. Rugplot
# Rugplot is especially plotted in pair with another plot like KDE or Hist which is uses to show ticks mainly in the graph

sns.kdeplot(
    data = tips,
    x="total_bill"
)
sns.rugplot(
    data = tips,
    x="total_bill"
)

#The orange lines are tips where high density has high ticks
```

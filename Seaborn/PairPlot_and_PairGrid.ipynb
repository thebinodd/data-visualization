```python
import seaborn as sns
```

```python
iris = sns.load_dataset("iris")
iris.sample(3)
```

```python
#-----Pair Plot-----

#Pair plot is used to plot every possible pair of numeric columns in the data
sns.pairplot(
    data = iris,
    hue="species"
)

# This has one rule :- the diagonal plots are either KDE or Histogram while non-diagonal diagram are always scatterplots
```

```python
# ----- Pair Grid -----

# This is just like pairplot but we have freedom over choosing plots

g = sns.PairGrid(data = iris , hue="species")
g.map_diag(sns.violinplot)
g.map_offdiag(sns.scatterplot)
```

```python
g = sns.PairGrid(data = iris , hue="species")
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```
